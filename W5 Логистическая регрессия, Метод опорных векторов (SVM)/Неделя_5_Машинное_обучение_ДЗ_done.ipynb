## Логистическая регрессия

Реализуй бинарную логистическую регрессию без регуляризации. Вычисляй градиенты сразу по всей выборке, без циклов по объектам. Цикл по итерациям градиентного спуска нужен. Затем проверь модель на Wine и построй границы для пар признаков. Проверки ниже охватывают основные свойства реализации, но не доказывают корректность на любых данных.

### Задание 1 [2 балла]

Напиши класс `LogisticRegression`. Используй среднюю бинарную кросс-энтропию и полный градиент по train. В `fit` заново инициализируй веса, очисти `losses` и сохраняй потерю после каждой итерации. Метод возвращает `self`. Для устойчивого вычисления потерь по логитам используй `np.logaddexp(0, z) - y*z`.

```python
class LogisticRegression:
    def __init__(self, lr=0.01, iterations=1000):
        self.lr = lr
        self.iterations = iterations
        self.w = None
        self.b = None
        self.losses = []

    def initialize_parameters(self, X):
        """
        Инициализирует веса и смещение нулями
        Принимает:
            X (ndarray, shape: [n_samples, n_features]): обучающие признаки
        На выходе:
            ничего не возвращает
        """
        self.w = np.zeros(X.shape[1])
        self.b = 0

    def _sigmoid(self, z):
        """
        Вычисляет сигмоидную функцию активации: 1 / (1 + e^(-z))
        Принимает:
            z (ndarray): входные значения
        Возвращает:
            sigmoid (ndarray): значения после сигмоиды
        """
        z = np.clip(z, -500, 500)  # Ограничиваем значения, чтобы избежать переполнения
        sigmoid = 1 / (1 + np.exp(-z))
        return sigmoid

    def compute_loss(self, X, y):
        """
        Вычисляет функцию потерь: бинарная кросс-энтропия
            L = -1/n * Σ(y*log(p) + (1-y)*log(1-p))
            Подсказка: используй np.mean(np.logaddexp(0, z) - y*z)
        Принимает:
            X (ndarray): массив признаков
            y (ndarray): вектор меток классов (0 или 1)
        Возвращает:
            loss (float): значение функции потерь
        """
        z = np.dot(X, self.w) + self.b
        loss = np.mean(np.logaddexp(0, z) - y * z)
        return loss

    def step(self, X, y):
        """
        Один шаг градиентного спуска по всей выборке

        Принимает:
            X (ndarray, shape: [n_samples, n_features]): обучающие признаки
            y (ndarray): метки классов (0 или 1)

        Возвращает:
            ничего не возвращает, обновляет веса и смещение
        """
        n = X.shape[0]
        z = np.dot(X, self.w) + self.b
        preds = self._sigmoid(z)          # предсказанные вероятности
        error = preds - y                 # ошибка: (p - y)

        dw = (1 / n) * np.dot(X.T, error)  # градиент по весам
        db = (1 / n) * np.sum(error)       # градиент по смещению

        self.w -= self.lr * dw             # обновление весов
        self.b -= self.lr * db             # обновление смещения

    def fit(self, X, y):
        """
        Обучает логистическую регрессию на тренировочных данных

        Принимает:
            X (ndarray, shape: [n_samples, n_features]): обучающие признаки
            y (ndarray): метки классов (0 или 1)

        Возвращает:
            self: обученный объект
        """
        self.losses = []
        self.initialize_parameters(X)
        for i in range(self.iterations):
            self.step(X, y)
            loss = self.compute_loss(X, y)
            self.losses.append(loss)
        return self

    def predict_proba(self, X):
        """
        Делает вероятностное предсказание

        Принимает:
            X (ndarray, shape: [n_samples, n_features]): обучающие признаки

        Возвращает:
            probs (ndarray): предсказанные вероятности классов
        """
        z = np.dot(X, self.w) + self.b
        probs = self._sigmoid(z)
        return np.vstack([1 - probs, probs]).T

    def predict(self, X):
        """
        Делает бинарное предсказание: класс 0 или 1

        Принимает:
            X (ndarray, shape: [n_samples, n_features]): обучающие признаки

        Возвращает:
            preds (ndarray): предсказанные метки классов
        """
        probs = self.predict_proba(X)[:, 1]
        preds = (probs >= 0.5).astype(int)
        return preds
```

Проверь формы массивов, значения сигмоиды и работу основных методов. Для обучения далее используем масштабированные признаки.

Построй границы для всех пар признаков Wine. Для каждой пары обучи `StandardScaler` на `X_train_raw[:, [i, j]]`, преобразуй train и обучи свою логистическую регрессию. Сетку создавай в исходных единицах и преобразуй тем же scaler перед `predict_proba`.

Используй `XX, YY = np.meshgrid(xx, yy)`, фон вероятностей со шкалой от 0 до 1 и линию уровня 0,5. На рисунке покажи объекты train. Разбей 78 пар на несколько рисунков, чтобы подписи оставались читаемыми. Это иллюстрация работы моделей на train, а не оценка качества на новых данных.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

def plot_logreg_boundaries_all_pairs(X_train_raw, y_train, feature_names, LogisticRegression):
    """
    Рисует границы решающего правила логистической регрессии для всех пар признаков Wine.
    Обучение на train-выборке, масштабирование через StandardScaler для каждой пары.
    """
    n_features = X_train_raw.shape[1]
    pairs = [(i, j) for i in range(n_features) for j in range(i + 1, n_features)]
    
    pairs_per_fig = 6
    n_figs = int(np.ceil(len(pairs) / pairs_per_fig))
    
    for fig_idx in range(n_figs):
        fig, axes = plt.subplots(2, 3, figsize=(15, 9))
        axes = axes.flatten()
        
        start = fig_idx * pairs_per_fig
        end = min(start + pairs_per_fig, len(pairs))
        
        for k, pair_idx in enumerate(range(start, end)):
            i, j = pairs[pair_idx]
            ax = axes[k]
            
            X_pair = X_train_raw[:, [i, j]]
            
            scaler = StandardScaler()
            X_pair_scaled = scaler.fit_transform(X_pair)
            
            lr = LogisticRegression()
            lr.fit(X_pair_scaled, y_train)
            
            x_min, x_max = X_pair[:, 0].min() - 0.3, X_pair[:, 0].max() + 0.3
            y_min, y_max = X_pair[:, 1].min() - 0.3, X_pair[:, 1].max() + 0.3
            xx = np.linspace(x_min, x_max, 300)
            yy = np.linspace(y_min, y_max, 300)
            XX, YY = np.meshgrid(xx, yy)
            grid = np.c_[XX.ravel(), YY.ravel()]
            
            grid_scaled = scaler.transform(grid)
            probs = lr.predict_proba(grid_scaled)[:, 1]
            Z = probs.reshape(XX.shape)
            
            # Фон — нейтральная палитра (не конфликтует с цветами точек)
            contour = ax.contourf(XX, YY, Z, levels=20, cmap='coolwarm', vmin=0, vmax=1, alpha=0.4)
            
            ax.contour(XX, YY, Z, levels=[0.5], colors='gray', linewidths=1.5, linestyles='--')
            
            # Точки — контрастные цвета с жёсткой обводкой
            scatter = ax.scatter(X_pair[:, 0], X_pair[:, 1], c=y_train,
                                 cmap='tab10', edgecolors='black', linewidths=1.2, s=45, zorder=3)
            
            ax.set_xlabel(feature_names[i])
            ax.set_ylabel(feature_names[j])
            ax.set_title(f'{feature_names[i]} vs {feature_names[j]}')
            ax.set_xlim(x_min, x_max)
            ax.set_ylim(y_min, y_max)
        
        for k in range(end - start, len(axes)):
            axes[k].axis('off')
        
        fig.colorbar(contour, ax=axes[:end - start], label='P(класс=1)', shrink=0.8)
        
        fig.suptitle(f'Границы логистической регрессии на train (Wine), часть {fig_idx + 1}/{n_figs}')
        plt.tight_layout()
        plt.show()
```

### Задание 4.1 [0,2 балла]

Заполни тело функции `apply_transformations()`. В функции уже реализована базовая версия: «Оригинал» — без преобразования и `log1p(y)` — лог-преобразование. Необходимо добавить ещё два преобразования:
- квадратный корень для неотрицательных цен. При обратном преобразовании отрицательный прогноз в шкале корня ограничь нулём, затем возведи в квадрат.

- Yeo — Johnson, более универсальное преобразование, похожее на Box — Cox, но умеет работать с отрицательными значениями и нулями, для его реализации используй `sklearn.preprocessing.PowerTransformer(method="yeo-johnson")`.

Функция должна возвращать словарь transforms:
- где ключ — название преобразования (строка);
- значение — кортеж из трёх элементов:
   - `y_transformed` — преобразованный `y_train`;
   - `inverse_transform_func` — функция, выполняющая обратное преобразование;
   - `transform_func` — функция, применяющая преобразование (она позволяет проверить прямое и обратное преобразования).

```python
def apply_transformations(y_train):
    y_train = np.asarray(y_train)
    transforms = {
        "Оригинал": (y_train, lambda x: x, lambda x: x),
        "log1p(y)": (np.log1p(y_train), np.expm1, np.log1p)
    }

    # Квадратный корень для неотрицательных цен. При обратном преобразовании отрицательный прогноз в шкале корня ограничь нулём, затем возведи в квадрат.
    sqrt_transform = np.sqrt(y_train)
    def inverse_sqrt(x):
        return np.maximum(x, 0) ** 2
    transforms["sqrt(y)"] = (sqrt_transform, inverse_sqrt, np.sqrt)


    # Yeo — Johnson, более универсальное преобразование, похожее на Box — Cox, но умеет работать с отрицательными значениями и нулями, для его реализации используй `sklearn.preprocessing.PowerTransformer(method="yeo-johnson")`
    yeo = PowerTransformer(method="yeo-johnson")
    y_yeo = yeo.fit_transform(y_train.reshape(-1, 1)).ravel()
    def inverse_yeo(x):
        return yeo.inverse_transform(np.asarray(x).reshape(-1, 1)).ravel()
    def transform_yeo(x):
        return yeo.transform(np.asarray(x).reshape(-1, 1)).ravel()
    transforms["yeo-johnson"] = (y_yeo, inverse_yeo, transform_yeo)

    """
    Функция должна возвращать словарь transforms:
    - где ключ — название преобразования (строка);
    - значение — кортеж из трёх элементов:
    - `y_transformed` — преобразованный `y_train`;
    - `inverse_transform_func` — функция, выполняющая обратное преобразование;
    - `transform_func` — функция, применяющая преобразование (она позволяет проверить прямое и обратное преобразования).
    """    
    
    return transforms
```

```python

def evaluate_model_cv(transform_name, config, X_train, y_train, X_test, y_test):
    
    #преобразователь таргета
    
    if transform_name == "Оригинал":
        transformer = FunctionTransformer(func=lambda x: x, inverse_func=lambda x: x)
    elif transform_name == "log1p(y)":
        transformer = FunctionTransformer(func=np.log1p, inverse_func=np.expm1)
    elif transform_name == "sqrt(y)":
        def inverse_sqrt(x):
            return np.maximum(x, 0) ** 2
        transformer = FunctionTransformer(func=np.sqrt, inverse_func=inverse_sqrt)
    elif transform_name == "yeo-johnson":
        transformer = PowerTransformer(method="yeo-johnson")
    else:
        raise ValueError(f"Неизвестное преобразование: {transform_name}")

    # 2. Собираем pipeline и оборачиваем в TransformedTargetRegressor
    pipeline = Pipeline([
        ("preprocess", clone(preprocessor)), 
        ("model", clone(config["estimator"]))
    ])

    model = TransformedTargetRegressor(
        regressor=pipeline,
        transformer=transformer
    )

    # 3. Единый KFold для всех вариантов
    kf = KFold(n_splits=5, shuffle=True, random_state=42)

    # 4. GridSearchCV
    grid = GridSearchCV(
        model,
        config["param_grid"],
        cv=kf,
        scoring="neg_mean_absolute_error",
        n_jobs=-1
    )

    grid.fit(X_train, y_train)

    # 5. Метрики
    cv_mae = -grid.best_score_  # neg_mean_absolute_error -> берём модуль

    y_pred = grid.predict(X_test)
    test_mae = mean_absolute_error(y_test, y_pred)
    test_r2 = r2_score(y_test, y_pred)

    return {
        "best_params": grid.best_params_,
        "cv_mae": cv_mae,
        "test_mae": test_mae,
        "test_r2": test_r2
    }   
```

Реализуй калибровку Платта для бинарных меток 0/1 или −1/+1. Используй

$$p_i=\frac{1}{1+\exp(Af_i+B)},\qquad L=-\sum_i[t_i\log p_i+(1-t_i)\log(1-p_i)].$$

Здесь $f_i$ является оценкой базовой модели на объектах, которые не участвовали в её обучении. Сглаженные цели равны $(N_++1)/(N_++2)$ для положительных меток и $1/(N_-+2)$ для отрицательных. $N_+$ и $N_-$ являются числом положительных и отрицательных меток калибровочной выборки.

Минимизируй отрицательное логарифмическое правдоподобие по A и B. Для устойчивости используй `np.logaddexp`, проверь успешность оптимизации. Подробности: [Platt scaling и численная реализация](https://www.csie.ntu.edu.tw/~cjlin/papers/plattprob.pdf).

```python
from scipy.optimize import minimize

class PlattCalibration:
    def __init__(self):
        """
        Инициализирует параметры A и B
        Эти параметры подбираются по минимуму отрицательного логарифмического правдоподобия
        """
        self.A = None
        self.B = None

    def _calc_t(self, y_array):
        """
        Вычисляет сглаженные целевые вероятности по числу меток каждого класса.

        Принимает:
            y_array (ndarray): массив меток классов

        Возвращает:
            t_array (ndarray): массив целевых вероятностей t
        """
        # Напиши код здесь
        return t_array

    def _sigmoid(self, f, A, B):
        """
        Вычисляет сигмоидную функцию с параметрами A и B:
            p = 1 / (1 + exp(A * f + B))

        Принимает:
            f (ndarray): выходы модели (decision_function)
            A (float): параметр наклона
            B (float): параметр смещения

        Возвращает:
            p_array (ndarray): вероятности класса 1
        """
        # Напиши код здесь
        return p_array

    def _neg_log_likelihood(self, params, f, y_array):
        """
        Вычисляет отрицательное логарифмическое правдоподобие (функцию потерь):
            L = -Σ[t * log(p) + (1 - t) * log(1 - p)]

        Принимает:
            params (list): список из двух элементов [A, B]
            f (ndarray): предсказания модели (decision function)
            y_array (ndarray): истинные метки классов (0/1 или -1/1)

        Возвращает:
            loss (float): значение функции потерь
        """
        # Напиши код здесь
        return loss

    def fit(self, f, y_array):
        """
        Обучает параметры A и B методом минимизации отрицательного логарифмического правдоподобия.

        Принимает:
            f (ndarray): выходы модели
            y_array (ndarray): метки классов (0/1 или -1/1)

        Возвращает:
            self (PlattCalibration): обученный объект
        """
        # Напиши код здесь
        return self

    def predict_proba(self, f):
        """
        Возвращает откалиброванные вероятности принадлежности к классу 1.

        Принимает:
            f (ndarray): предсказания модели (например, decision_function)

        Возвращает:
            probs (ndarray): откалиброванные вероятности класса 1
        """
        # Напиши код здесь
        return probs
```
