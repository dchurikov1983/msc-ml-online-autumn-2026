```python
def compare_rmse(rmse_tuning, rmse_baseline):
    """Относительное изменение RMSE по сравнению с базовой моделью."""
    if rmse_baseline <= 0:
        raise ValueError('Для относительного сравнения нужен положительный baseline RMSE')
    improvement = 100 * (rmse_baseline - rmse_tuning) / rmse_baseline
    print(f'Изменение RMSE относительно baseline: {improvement:+.2f}% улучшения')
    return improvement
```
